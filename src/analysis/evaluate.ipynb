## R 9346-I/109

```python
# dateien korrekt oeffnen
with open("../../data/groundtruth/02_ocr/corrected_R 9346-I_109_ocr.json", "r", encoding="utf-8") as f:
    ref_data = json.load(f)

with open("../../data/groundtruth/02_ocr/R 9346-I_109_ocr.json", "r", encoding="utf-8") as f:
    hyp_data = json.load(f)

# string aus json-struktur extrahieren (key anpassen, falls abweichend)
ref_text = ref_data["text"] if isinstance(ref_data, dict) else str(ref_data)
hyp_text = hyp_data["text"] if isinstance(hyp_data, dict) else str(hyp_data)

# cer berechnen
error = jiwer.cer(ref_text, hyp_text)
print(f"CER: {error:.4f}")
```

## nur transcription (einzeln)

```python
# dateien laden
with open("../../data/groundtruth/02_ocr/corrected_R 9346-I_109_ocr.json", "r", encoding="utf-8") as f:
    ref_json = json.load(f)

with open("../../data/groundtruth/02_ocr/R 9346-I_109_ocr.json", "r", encoding="utf-8") as f:
    hyp_json = json.load(f)

# cer pro seite/transkription berechnen
for idx, (ref_item, hyp_item) in enumerate(zip(ref_json, hyp_json), start=1):
    ref_text = ref_item.get("transcription", "")
    hyp_text = hyp_item.get("transcription", "")
    
    # leere transkriptionen abfangen
    if not ref_text and not hyp_text:
        print(f"Seite {idx}: Keine Transkription vorhanden (übersprungen)")
        continue
        
    page_cer = jiwer.cer(ref_text, hyp_text)
    print(f"Seite {idx} ({ref_item.get('_filename', 'Unbekannt')}): CER = {page_cer:.4f} ({page_cer * 100:.2f}%)")
```
